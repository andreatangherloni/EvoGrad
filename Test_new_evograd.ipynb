```python
from evograd.core.problem import Problem
from evograd.core.minimize import minimize
from evograd.core.maximize import maximize

from evograd.core.termination import MaxEvaluations
from evograd.operators import TournamentSelection, SBXCrossover, PolynomialMutation, ReflectRepair
from evograd.algorithms.ga import GA
from evograd.algorithms.de import DE
from evograd.algorithms.pso import PSO
from evograd.algorithms.cmaes import CMAES
from evograd.algorithms.shade import SHADE, LSHADE
```

```python
problem = Problem(
    objective=lambda x: (x ** 2).sum(dim=-1),  # Sphere function
    n_var=10,
    xl=-5.0,
    xu=5.0,
)
```

# Test GA differentiability

```python
algorithm = GA(pop_size=100,
               selection=TournamentSelection(tournament_size=3, differentiable=False),
               crossover=SBXCrossover(eta=15, prob=0.9, differentiable=False),
               mutation=PolynomialMutation(eta=20, differentiable=False),
               differentiable=False,
               )

result = minimize(problem,
                  algorithm,
                  termination=MaxEvaluations(1000),
                  seed=42,
                  verbose=False)

print("Diff pop?", algorithm.population.requires_grad)
print("Diff sl? ", algorithm.selection.temperature.requires_grad)
print("Diff cr? ", algorithm.crossover.eta.requires_grad)
print("Diff mt? ", algorithm.mutation.eta.requires_grad)
```
